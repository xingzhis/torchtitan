```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python
# 1. Define the path to your checkpoint
model_path = "/gpfs/gibbs/pi/krishnaswamy_smita/xingzhi/Transformer-Dispersion/pretrain_qwen3_0.6B_ckpts/step-29610"

print(f"Loading model from {model_path}...")

# 2. Load the tokenizer and model
# trust_remote_code=True is often required for Qwen models
tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    model_path,
    device_map="auto",          # Automatically chooses GPU/CPU
    torch_dtype=torch.bfloat16, # Recommended for Qwen architectures
    trust_remote_code=True
)
```

```python

# 3. Prepare your test prompt
prompt = "The future of artificial intelligence is"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# 4. Generate text
print("\n--- Generating Output ---")
output_tokens = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.7,
    top_p=0.9,
    repetition_penalty=1.1
)

# 5. Decode and print
generated_text = tokenizer.decode(output_tokens[0], skip_special_tokens=True)
print(generated_text)
```

```python
import torch
from transformers import AutoModelForCausalLM

model_path = "/gpfs/gibbs/pi/krishnaswamy_smita/xingzhi/Transformer-Dispersion/pretrain_qwen3_0.6B_ckpts/step-29610"
model = AutoModelForCausalLM.from_pretrained(model_path, torch_dtype=torch.bfloat16)

# 1. Check for "Zero" or "Dead" weights
for name, param in model.named_parameters():
    if param.abs().max() < 1e-8:
        print(f"⚠️ WARNING: {name} appears to be empty/zero!")
    if torch.isnan(param).any():
        print(f"❌ CRITICAL: {name} contains NaNs!")

# 2. Compare Embeddings vs LM Head
# In many Qwen/Llama models, these weights are tied or should have similar scales
embed_weight = model.get_input_embeddings().weight
lm_head_weight = model.get_output_embeddings().weight

print(f"Embedding Mean: {embed_weight.mean().item():.6f}, Std: {embed_weight.std().item():.6f}")
print(f"LM Head Mean: {lm_head_weight.mean().item():.6f}, Std: {lm_head_weight.std().item():.6f}")

# If Std is very high (e.g., > 1.0) or mean is exactly 0.0, the conversion failed.
```
